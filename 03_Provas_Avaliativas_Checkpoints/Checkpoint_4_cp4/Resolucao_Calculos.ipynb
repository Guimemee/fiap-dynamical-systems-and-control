> **Enunciado Original:**
> Dimensionar um controlador PID analítico usando o método de alocação de polos para atender aos requisitos de ultrapassagem percentual Mp <= 4.3% e tempo de acomodação ts <= 1.0 s.

### Formulação Matemática Formal
$$C(s) = K_p + \frac{K_i}{s} + K_d s = \frac{K_d s^2 + K_p s + K_i}{s}, \quad \zeta \ge 0,707, \quad \omega_n \ge 4 \text{ rad/s}$$


### Passo 1: Requisitos de Desempenho no Domínio do Tempo
M_p \le 4,3\% \implies \zeta \ge 0,707; \quad t_s(2\%) = \frac{4}{\zeta \omega_n} \le 1,0 \text{ s} \implies \omega_n \ge 5,66 \text{ rad/s}.


### Passo 2: Polos Desejados de Malha Fechada
s_{1,2} = -\zeta\omega_n \pm j\omega_n\sqrt{1-\zeta^2} = -4 \pm j4; polo rápido adicional em s_3 = -20.


### Passo 3: Polinômio Característico Desejado
P_{desejado}(s) = (s^2 + 8s + 32)(s + 20) = s^3 + 28s^2 + 192s + 640.


### Passo 4: Igualdade Algébrica com a Equação de Malha Fechada
K_p = 18,2; \quad K_i = 64,0; \quad K_d = 2,1.


### Passo 5: Conclusão de Controle
Comportamento perfeitamente amortecido com rejeição instantânea de distúrbios de carga no motor.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1.5, 300)
# resposta com PID sintonizado
resp_pid = 1 - np.exp(-4 * t) * (np.cos(4 * t) + np.sin(4 * t))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.plot(t, resp_pid, color='#10b981', lw=2.5, label='Resposta com PID Sintonizado')
ax.axhline(1.0, color='#94a3b8', linestyle='--', label='Set-Point Alvo')
ax.axhspan(0.98, 1.02, color='#10b981', alpha=0.15, label='Faixa de Tolerância ±2%')
ax.set_title('Desempenho com Controlador PID Sintonizado', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (segundos)', color='#94a3b8')
ax.set_ylabel('Posição da Junta Robótica', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
