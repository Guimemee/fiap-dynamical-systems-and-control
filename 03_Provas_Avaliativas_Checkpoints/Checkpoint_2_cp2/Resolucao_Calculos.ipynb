> Simplificar o diagrama de blocos multimalha com realimentação unitária e perturbações externas determinando a função de transferência de malha fechada e sensibilidade a ruídos.

### Formulação Matemática Formal
$$T(s) = \frac{C(s)}{R(s)} = \frac{G_c(s) G_p(s)}{1 + G_c(s) G_p(s) H(s)}, \quad S(s) = \frac{1}{1 + G_c(s) G_p(s)}$$


### Passo 1: Identificação de Laços de Realimentação e Blocos em Cascata
Produto direto de funções em série e divisão por (1 + GH) para laços internos.


### Passo 2: Movimentação de Pontos de Soma e Ramificação
Equivalência matemática mantendo os sinais algébricos inalterados.


### Passo 3: Função de Transferência de Malha Fechada
T(s) = \frac{100}{s^2 + 14,14s + 100} com \omega_n = 10 \text{ rad/s} e \zeta = 0,707.


### Passo 4: Função de Sensibilidade
S(s) demonstra atenuação de perturbações de baixa frequência superior a 40 dB.


### Passo 5: Conclusão de Projeto
Erro em regime permanente a entrada em degrau identicamente nulo e transitório suave.


```python
import numpy as np
import matplotlib.pyplot as plt

w = np.logspace(-1, 2, 100)
sensibilidade_db = 20 * np.log10(np.abs(w**2 / (100 - w**2 + 14.14j*w)))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.semilogx(w, sensibilidade_db, color='#06b6d4', lw=2.5, label='Sensibilidade S(jω) dB')
ax.set_title('Atenuação de Perturbações (Função de Sensibilidade)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Frequência (rad/s)', color='#94a3b8')
ax.set_ylabel('Magnitude de Sensibilidade (dB)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#06b6d4', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
