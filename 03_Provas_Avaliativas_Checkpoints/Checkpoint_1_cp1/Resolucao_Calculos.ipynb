> **Enunciado Original:**
> Obter a função de transferência G(s) de um braço robótico de precisão acionado por motor DC considerando inércia J, atrito viscoso b, resistência R e indutância L.

### Formulação Matemática Formal
$$G(s) = \frac{\Theta(s)}{V(s)} = \frac{K_t}{s \left[ (J s + b)(L s + R) + K_t K_e \right]}$$


### Passo 1: Equação Elétrica da Armadura
V(t) = R i(t) + L \frac{di(t)}{dt} + e_b(t), \text{ onde } e_b(t) = K_e \omega(t).


### Passo 2: Equação Mecânica de Torque e Rotação
T(t) = J \frac{d\omega(t)}{dt} + b \omega(t), \text{ onde } T(t) = K_t i(t).


### Passo 3: Aplicação da Transformada de Laplace com Condições Iniciais Nulas
V(s) = (Ls + R)I(s) + K_e s\Theta(s) \quad \text{e} \quad K_t I(s) = (Js^2 + bs)\Theta(s).


### Passo 4: Eliminação da Corrente I(s)
\frac{\Theta(s)}{V(s)} = \frac{K_t}{JL s^3 + (JR + bL)s^2 + (bR + K_t K_e)s}.


### Passo 5: Conclusão Dinâmica
Como L \ll R, o sistema simplifica para modelo de segunda ordem padrão com polos estáveis no semiplano esquerdo.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2, 200)
# resposta ao degrau de segunda ordem
wn = 10
zeta = 0.707
y = 1 - (np.exp(-zeta * wn * t) / np.sqrt(1 - zeta**2)) * np.sin(wn * np.sqrt(1 - zeta**2) * t + np.arccos(zeta))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.plot(t, y, color='#10b981', lw=2.5, label='Posição Angular θ(t)')
ax.axhline(1.0, color='#94a3b8', linestyle='--', label='Referência de Posição')
ax.set_title('Resposta ao Degrau Unitário do Braço Robótico', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (segundos)', color='#94a3b8')
ax.set_ylabel('Posição Normalizada', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
