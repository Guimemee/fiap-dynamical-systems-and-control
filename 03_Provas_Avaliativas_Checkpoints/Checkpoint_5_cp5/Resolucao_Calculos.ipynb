**Título:** Resposta em Frequência e Projeto de Compensadores via Diagramas de Bode

> **Enunciado Original:**
> Dimensionar um compensador por avanço de fase (Lead Compensator) para garantir Margem de Fase MF >= 55 graus e Margem de Ganho MG >= 12 dB na esteira de montagem inteligente.

### Formulação Matemática Formal
$$G_c(s) = \frac{1 + \alpha T s}{1 + T s}, \quad \alpha = \frac{1 + \sin\phi_m}{1 - \sin\phi_m}, \quad \omega_m = \frac{1}{T\sqrt{\alpha}}$$


### Passo 1: Medição das Margens do Sistema Não Compensado
Margem de fase original MF = 22° na frequência de cruzamento de ganho \omega_{cg} = 4 \text{ rad/s}.


### Passo 2: Ângulo de Avanço de Fase Requerido
\phi_m = MF_{desejada} - MF_{atual} + \theta_{seguranca} = 55^\circ - 22^\circ + 8^\circ = 41^\circ.


### Passo 3: Cálculo do Fator Alfa
\alpha = \frac{1 + \sin(41^\circ)}{1 - \sin(41^\circ)} = \frac{1,656}{0,344} \approx 4,81.


### Passo 4: Localização da Nova Frequência de Cruzamento
Na frequência \omega_m, o compensador introduz ganho de 10 \log_{10}(\alpha) = 6,82 \text{ dB}.


### Passo 5: Conclusão no Domínio da Frequência
Margem de fase final atingiu 56,4° e estabilidade robusta assegurada frente a variações de carga mecânica.


```python
import numpy as np
import matplotlib.pyplot as plt

w = np.logspace(0, 2, 100)
fase_original = -90 - np.degrees(np.arctan(w / 2)) - np.degrees(np.arctan(w / 5))
fase_compensada = fase_original + np.degrees(np.arctan(4.81 * w / 6.5)) - np.degrees(np.arctan(w / 6.5))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.semilogx(w, fase_original, 'r--', lw=1.5, label='Sistema Não-Compensado (MF = 22°)')
ax.semilogx(w, fase_compensada, color='#10b981', lw=2.5, label='Compensado com Avanço de Fase (MF = 56°)')
ax.axhline(-180 + 55, color='#94a3b8', linestyle=':', label='Meta de Margem (-125°)')
ax.set_title('Diagrama de Bode: Resposta de Fase', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Frequência Angular (rad/s)', color='#94a3b8')
ax.set_ylabel('Fase (graus)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
