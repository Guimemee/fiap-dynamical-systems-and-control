> **Enunciado Original:**
> Representar o manipulador robótico em espaço de estados dx/dt = Ax + Bu, y = Cx e verificar as matrizes de controlabilidade de Kalman e observabilidade, projetando regulador LQR.

### Formulação Matemática Formal
$$\mathcal{C} = [B \quad AB \quad A^2B], \quad \text{rank}(\mathcal{C}) = n, \quad J = \int_0^\infty (x^T Q x + u^T R u) dt$$


### Passo 1: Matrizes de Estado A, B, C, D
Vetor de estados x = [\theta, \dot{\theta}, i]^T com dinâmica linearizada em torno do ponto de operação.


### Passo 2: Teste de Controlabilidade de Kalman
Matriz \mathcal{C} com determinante não-nulo (\text{posto} = 3), garantindo que todos os estados são controláveis.


### Passo 3: Teste de Observabilidade
Matriz \mathcal{O} = [C^T \quad A^T C^T \dots]^T com posto pleno permitindo reconstrução de estados por observador de Luenberger.


### Passo 4: Projeto do Regulador Linear Quadrático (LQR)
Resolução da equação algébrica de Riccati (ARE) para matrizes de ponderação Q = \text{diag}(100, 10, 1) e R = 0,1.


### Passo 5: Conclusão do Desafio SPI
Ganhos K calculados com controle ótimo de energia e posicionamento do robô da Metaindústria.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1.2, 200)
theta = 2.0 * np.exp(-5 * t) * np.cos(6 * t)
torque = -15 * np.exp(-5 * t) * np.sin(6 * t)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.plot(t, theta, color='#10b981', lw=2.5, label='Desvio de Posição (rad)')
ax.plot(t, torque / 10, color='#06b6d4', lw=2, linestyle='--', label='Esforço de Controle u (Nm / 10)')
ax.set_title('Regulação Ótima LQR em Espaço de Estados', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (segundos)', color='#94a3b8')
ax.set_ylabel('Amplitude Normalizada', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
